```python
#1: Count Vowels

line=input("Enter a line: ").lower()
count=0
for c in "aeiou":
    count+=line.count(c)
print(count)
```

```python
#2: Reverse String

word=input("Enter word:")
rword=""
for i in range(len(word)-1,-1,-1):
    rword+=word[i]
print(rword)
```

```python
#6: Title case

line1=input("Enter a line: ").split()
line2=""
for word in line1:
    line2+=word[0].upper()+word[1:].lower()+" "
print(line2)
```

```python
#7: Toggle case

line1=input("Enter a line: ")
line2=""
for word in line1:
    if word.isupper():
      word=word.lower()
      line2+=word
    else:
      word=word.upper()
      line2+=word

print(line2)
```

```python
#8: Caesar Cipher

msg=input("String: ")
shift=int(input("Shift: "))
encrypted=""
for char in msg:
    if char>='a' and char<='z':
        encrypted_data=(ord(char)-ord('a')+shift)%26+ord('a')
        encrypted_character=chr(encrypted_data)
        encrypted+=encrypted_character
    else:
        encrypted+=char#spaces
print(encrypted)
```

```python
#9: Remove Duplicates

ip=input("Enter a string: ")
op=""
for i in ip:
  if i not in op:
    op+=i
print(op)
```

```python
#10: Count Character Frequency

d={}
a=input("Enter a string: ")
for i in a:
  count=1
  if i not in d:
    d[i]=count
  else:
    count+=1
    d[i]=count

print(d)
```

```python
#11: Find First Substring Index

string=input("String: ")
substring=input("Substring: ")
flag=-1
for i in range(len(string)-len(substring)+1):
    if string[i:i+len(substring)]==substring:
        flag=i
        break

print(flag)
```

```python
#12 Extract Middle Characters

a = input("Enter String: ")
length=len(a)
middle=length//2
if length%2==0:
  ch1=a[middle-1]
  ch2=a[middle]
  print(f"{ch1},{ch2}")
elif length%2!=0:
  print(a[middle])
```

```python
#13 Remove Nth index

a,index=input("String: "),int(input("Enter index"))
new=""
for i,ch in enumerate(a):
  if i!=index:
    new=new+ch

print(new)
```

```python
#15 Longest Word

sen=input("Enter a sentence: ")
length = 0
max_length = 0

for ch in sen:
    if ch!=" ":
        length+=1
    else:
        if length>max_length:
            max_length=length
        length=0

if length>max_length:
    max_length=length

print(max_length)
```

```python
#16: Check for only digits

s=input("Enter a string: ")
only_digits=True
for ch in s:
    if ch<'0' or ch>'9':
        only_digits=False
        break

if only_digits:
    print(True)
else:
    print(False)

```

```python
#17: String Compression

line=input("Enter a line: ")
compressed_line=""
i=j=0
while i<len(line):
    alpha=line[i]
    j=i+1
    while j<len(line) and line[i]==line[j]:
        j+=1
    compressed_line+=alpha+str(j-i)
    i=j

print(compressed_line)
```

```python
#18: Check Prefix and Suffix

s=input("Enter the string: ")
prefix=input("Enter prefix: ")
suffix=input("Enter suffix: ")

is_prefix = True
for i in range(len(prefix)):
    if s[i]!=prefix[i]:
        is_prefix=False
        break

is_suffix = True
for i in range(len(suffix)):
    if s[len(s)-len(suffix)+i]!=suffix[i]:
        is_suffix=False
        break

if is_prefix and is_suffix:
    print(True)
else:
    print(False)
```

```python
#19: Interleave Strings

str1=input("Enter first string: ")
str2=input("Enter second string: ")
result=""
i=0
j=0
while i<len(str1) and j<len(str2):
    result+=str1[i]
    result+=str2[j]
    i+=1
    j+=1

while i<len(str1):
    result+=str1[i]
    i+=1

while j<len(str2):
    result+=str2[j]
    j+=1

print(result)
```
