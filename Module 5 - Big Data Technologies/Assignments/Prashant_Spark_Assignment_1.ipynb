```python
sc = spark.sparkContext
```

```python
rdd1=sc.textFile('/content/sample_data/airlines.csv')
```

```python
for var in rdd1.take(5):
  print(var)
```

```python
header=rdd1.first()
```

```python
rdd2=rdd1.filter(lambda a : a!=header)
```

```python
for var in rdd2.collect():
  print(var)
```

```python
rdd3=rdd2.map(lambda a : (a.split(",")[0],float(a.split(",")[2])))
```

```python
for var in rdd3.collect():
  print(var)
```

```python
rdd4=rdd3.reduceByKey(lambda a,b : round((a+b)/2,2))
```

```python
for var in rdd4.collect():
    print(var)
```

```python
airlines_DF = spark.read.format("csv").option("header","true").option("inferSchema","true").load("/content/sample_data/airlines.csv")
airlines_DF.printSchema()
```

```python
airlines_DF.createOrReplaceTempView("airlines")
```

*Q2*

```python
q1=spark.sql("select Year, round(avg(Avg_rev_per_seat),2) as Average_Revenue_Per_Seat from airlines group by Year")
q1.show()
```

```python
q2=spark.sql("Select Quarter, round(avg(Avg_rev_per_seat),2) as Average_Revenue_Per_Seat from airlines group by Quarter")
q2.show()
```

*Q3*

```python
q3=spark.sql("Select Year, round(avg(Avg_rev_per_seat),2) as Highest_Average_Revenue_Per_Seat from airlines group by Year order by Highest_Average_Revenue_Per_Seat desc Limit 1")
q3.show()
```

```python
q4=spark.sql("Select Quarter, round(avg(Avg_rev_per_seat),2) as Highest_Average_Revenue_Per_Seat from airlines group by Quarter order by Highest_Average_Revenue_Per_Seat desc Limit 1")
q4.show()
```
