#**Ques3**

```python
rdd1 = sc.textFile("/content/sample_data/Data/txn.txt")
```

```python
from datetime import datetime
```

```python
format_date = "%m-%d-%Y"
```

```python
rdd2 = rdd1.map(lambda a : (datetime.strptime(a.split(',')[1], format_date).strftime("%w") + " " + (datetime.strptime(a.split(',')[1], format_date).strftime("%A")), float(a.split(",")[3])))
```

```python
for var in rdd2.take(5):
  print(var)
```

```python
rdd3 = rdd2.reduceByKey(lambda a,b : round(a + b, 2))
```

```python
rdd3.count()
```

```python
for var in rdd3.collect():
  print(var)
```

```python
rdd4 = rdd3.sortByKey()
```

```python
for var in rdd4.collect():
  print(var)
```

```python
csv_output = rdd4.map(lambda a : a[0].split(' ')[1] + ',' + str(round(a[1],2)))
```

```python
for var in csv_output.collect():
  print(var)
```

```python
csv = csv_output.coalesce(1)
```

```python
csv.getNumPartitions()
```

```python
csv.saveAsTextFile('/content/sample_data/Data/Day_wise_sales')
```

```python
rdd1 = sc.textFile("/content/sample_data/Data/txn.txt")
```

```python
rdd1.count()
```

```python
rdd2 = rdd1.map(lambda a : (a.split(',')[2], float(a.split(',')[3])))
```

```python
for var in rdd2.take(5):
  print(var)
```

```python
rdd3 = rdd2.reduceByKey(lambda a,b : round(a+b,2))
```

```python
for var in rdd3.take(5):
  print(var)
```

```python
sorted_top_custid = rdd3.sortBy(lambda a : -a[1])
```

```python
for var in sorted_top_custid.take(10):
  print(var)
```

```python
top10_cust = sorted_top_custid.take(10)
```

```python
print(top10_cust)
```

```python
rdd4 = sc.parallelize(top10_cust,10)
```

```python
for var in rdd4.collect():
  print(var)
```

```python
rdd5 = sc.textFile("/content/sample_data/Data/custs.txt")
```

```python
rdd6 = rdd5.map(lambda a : (a.split(',')[0], a.split(',')[1] + " " + a.split(',')[2]))
```

```python
for var in rdd6.take(5):
  print(var)
```

```python
joinRDD = rdd6.join(rdd4)
```

```python
for var in joinRDD.collect():
  print(var)
```

```python
rdd7 = joinRDD.map(lambda a : (a[1][0], a[1][1]))
```

```python
for var in rdd7.collect():
  print(var)
```

```python
top_cust_sorted = rdd7.sortBy(lambda a : -a[1])
```

```python
for var in top_cust_sorted.collect():
  print(var)
```

```python
csv_output1 = top_cust_sorted.map(lambda a : a[0] + ',' +  str(round(a[1],2)))
```

```python
for var in csv_output1.collect():
  print(var)
```

```python
csv1 = csv_output1.coalesce(1)
```
