```python
class BankAccount:
    account_no=1000
    def __init__(self,name,gender,balance):
        BankAccount.account_no+=1
        self.__account_number=BankAccount.account_no
        self.__name=name
        self.__gender=gender
        self.__balance=float(balance)
    def get_account_number(self):
        return self.__account_number
    def _get_balance(self):
        return self.__balance
    def withdraw(self,with_amount,min_balance=500,fee=500):
        try:
            with_amount=float(with_amount)
        except:
            print("Invalid amount");return
        if with_amount<=0:
            print("Invalid withdrawal amount");return
        if with_amount>self.__balance:
            print("Insufficient Funds");return
        if self.__balance-with_amount<min_balance:
            total=with_amount+fee
            if self.__balance-total<0:
                print("ERROR: Transaction denied. (Withdrawal + Fee would result in negative balance)");return
            self.__balance-=total
            print(f"SUCCESS: Withdrew {with_amount}. Fee {fee} applied. New balance: {self.__balance}")
        else:
            self.__balance-=with_amount
            print(f"SUCCESS: Withdrew {with_amount}. New balance: {self.__balance}")
    def deposit(self,dep_amount):
        try:
            dep_amount=float(dep_amount)
        except:
            print("Invalid amount");return
        if dep_amount<=0:
            print("Amount cannot be negative or zero");return
        self.__balance+=dep_amount
        print(f"SUCCESS: Deposited {dep_amount}. New balance: {self.__balance}")
    def show_account_info(self):
        print(f"Name:{self.__name}\tGender:{self.__gender}\tA/C No:{self.__account_number}\tBalance:{self.__balance}")
    def transfer_money(self,amount,other_account,min_balance=500,fee=500):
        try:
            amount=float(amount)
        except:
            print("Invalid amount");return
        if amount<=0:
            print("Invalid transfer amount");return
        if amount>self.__balance:
            print("Insufficient Funds");return
        if self.__balance-amount<min_balance:
            total=amount+fee
            if self.__balance-total<0:
                print("ERROR: Transaction denied. (Transfer + Fee would result in negative balance)");return
            self.__balance-=total
            other_account._BankAccount__balance+=amount
            print(f"SUCCESS: Transferred {amount} to {other_account.get_account_number()}. Fee {fee} applied. Your new balance: {self.__balance}")
        else:
            self.__balance-=amount
            other_account._BankAccount__balance+=amount
            print(f"SUCCESS: Transferred {amount} to {other_account.get_account_number()}. Your new balance: {self.__balance}")
    def __str__(self):
        return f"Name:{self.__name}\tGender:{self.__gender}\tA/C No:{self.__account_number}\tBalance:{self.__balance}"

bank={}
while True:
    print("Welcome to Bank Account")
    print("1: Create account")
    print("2: Deposit")
    print("3: Withdraw")
    print("4: Show account info")
    print("5: Transfer money")
    print("0: Exit")
    try:
        choice=int(input("Enter your choice:"))
    except:
        print("Invalid choice");continue
    if choice==1:
        name=input("Enter your name: ")
        gender=input("Enter your gender: ")
        try:
            bal=float(input("Enter balance:"))
        except:
            print("Invalid balance");continue
        if bal>=3000:
            acc=BankAccount(name,gender,bal)
            bank[acc.get_account_number()]=acc
            print("Account created successfully")
            acc.show_account_info()
        else:
            print("Amount should be at least 3000")
    elif choice==2:
        try:
            acc_no=int(input("Enter account number:"))
        except:
            print("Invalid account number");continue
        acc=bank.get(acc_no)
        if not acc:
            print("Account not found");continue
        try:
            dep_amount=float(input("Enter amount to deposit:"))
        except:
            print("Invalid amount");continue
        acc.deposit(dep_amount)
    elif choice==3:
        try:
            acc_no=int(input("Enter account number:"))
        except:
            print("Invalid account number");continue
        acc=bank.get(acc_no)
        if not acc:
            print("Account not found");continue
        try:
            with_amount=float(input("Enter amount to withdraw:"))
        except:
            print("Invalid amount");continue
        acc.withdraw(with_amount)
    elif choice==4:
        if not bank:
            print("No accounts found");continue
        for acc in bank.values():
            acc.show_account_info()
    elif choice==5:
        try:
            s_no=int(input("Enter sender account number:"))
            r_no=int(input("Enter receiver account number:"))
        except:
            print("Invalid account number");continue
        sender=bank.get(s_no);receiver=bank.get(r_no)
        if not sender or not receiver:
            print("Invalid account number(s)");continue
        try:
            amt=float(input("Enter amount to transfer:"))
        except:
            print("Invalid amount");continue
        sender.transfer_money(amt,receiver)
    elif choice==0:
        print("Exiting the system");break
    else:
        print("Invalid choice")
```
