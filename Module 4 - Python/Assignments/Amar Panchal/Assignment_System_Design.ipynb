```python
menu = {
    "tea": 15.00,
    "coffee": 25.00,
    "samosa": 20.00,
    "water": 20.00,
    "bun maska": 15.00
}
ordered_items=[]

def order_input():
    global ordered_items
    ordered_items = []
    order=input("Enter your order in the form of -->>> tea 1 coffee 2 ...): ").lower().strip()
    if not order:
        print("No order entered.")
        return

    order_parts=order.split()

    i=0
    order_dict={}
    while i<len(order_parts):
        name_parts=[]
        while i<len(order_parts) and not order_parts[i].isdigit():
            name_parts.append(order_parts[i])
            i+=1
        item_name=" ".join(name_parts).strip()
        if i<len(order_parts) and order_parts[i].isdigit():
            qty=int(order_parts[i])
            i+=1
        else:
            qty=1

        if not item_name:
            continue

        if item_name in menu:
            ordered_items.append((item_name,qty,menu[item_name]))
            order_dict[item_name]=qty
        else:
            print(f"Warning: '{item_name}' not on menu — skipped.")

    print(f"Your given order is--->{order_dict}")


def calculate_bill():
    if not ordered_items:
        print("No items ordered yet. Please place an order first.")
        return

    membership=input("Enter membership type (Gold/Silver/None): ").strip().lower()
    if membership=="gold":
        discount_rate=0.10
        membership_label="Gold (10.00% Off)"
    elif membership=="silver":
        discount_rate=0.05
        membership_label="Silver (5.00% Off)"
    else:
        discount_rate=0.00
        membership_label="None (0.00% Off)"

    subtotal=0.0
    for item in ordered_items:
        subtotal+=item[1]*item[2]

    subtotal=round(subtotal,2)

    discount=round(subtotal * discount_rate, 2)
    taxable_total=round(subtotal - discount, 2)

    gst=round(taxable_total * 0.05, 2)
    final_total=round(taxable_total + gst, 2)

    print("\n=========================================")
    print("             Apna Tea Stall                ")
    print("=========================================")
    print("ITEM                   QTY    PRICE    TOTAL")
    print("-----------------------------------------")
    for item in ordered_items:
        name=item[0].title().ljust(22)
        qty=str(item[1]).rjust(5)
        price=f"{item[2]:.2f}".rjust(8)
        total=f"{item[1]*item[2]:.2f}".rjust(8)
        print(f"{name}{qty}{price}{total}")
    print("-----------------------------------------")
    print(f"Subtotal{'':>29}{subtotal:.2f}")
    print(f"Membership: {membership_label}")
    print(f"Discount Applies{'':>20}-{discount:.2f}")
    print("-----------------------------------------")
    print(f"Taxable Amount{'':>24}{taxable_total:.2f}")
    print(f"GST (5.00%){'':>26}+{gst:.2f}")
    print("-----------------------------------------")
    print(f"Total Amount Due{'':>22}{final_total:.2f}")
    print("=========================================")
    print("Thank you! Please visit again.")
    ordered_items.clear()

while True:
    print("--------------")
    print("Apna Tea Stall")
    print("--------------")
    print("MENU CARD".center(15))
    print("--------------")
    print("* Tea * ".center(15))
    print("* Coffee *".center(15))
    print("* Samosa * ".center(15))
    print("* Water *".center(15))
    print("* Bun Maska *".center(15))
    print("* Exit *".center(15))
    print("--------------")
    print("Enter 1 to place order ")
    print("Enter 2 to calculate the bill")
    print("Enter 0 exit ")
    choice=int(input("Enter your choice: "))
    if choice==1:
        order_input()
    elif choice==2:
        calculate_bill()
    elif choice==0:
        print("Thanks for visit")
        break
    else:
        print("Please check the option.")
```
