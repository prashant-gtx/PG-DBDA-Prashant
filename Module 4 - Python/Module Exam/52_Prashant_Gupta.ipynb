```python
#Q1

def longest_increasing_subsequence(arr):
    n=len(arr)
    if n==0:
        return 0
    l=[1]*n
    for i in range(n):
        for j in range(i):
            if arr[j]<arr[i] and l[j]+1>l[i]:
                l[i]=l[j]+1
    return max(l)

def get_lis_elements(a):
    b=[]
    for i in a:
        for j in range(0,len(a)+1):
            if i<j and i not in b:
                b.append(i)
        for num in b:
            if num>i:
                b.remove(num)

    print(b)


#output: 
#longest_increasing_subsequence: 4
#count_lis: 2
#get_lis_elements: [2,3,7,18]
```

```python
arr=[10,9,2,5,3,7,101,18]
longest_increasing_subsequence(arr)
```
